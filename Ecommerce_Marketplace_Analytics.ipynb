## 6.5 Delivery Performance Against Estimated Date

Delivery performance is evaluated by comparing the actual customer delivery date with the estimated delivery date.

The classification is:

- **Early:** Actual delivery occurred before the estimated date.
- **On Time:** Actual delivery occurred on the estimated date.
- **Late:** Actual delivery occurred after the estimated date.

This metric helps evaluate whether the marketplace met its promised delivery expectations.

Extreme delays are retained in the analysis because they represent genuine observations in the source dataset. Their impact will be considered separately from the normal delivery pattern.

```python
print("Delivered before estimated date:",
      (delivery_analysis["delivery_delay_days"] < 0).sum())

print("Delivered on estimated date:",
      (delivery_analysis["delivery_delay_days"] == 0).sum())

print("Delivered after estimated date:",
      (delivery_analysis["delivery_delay_days"] > 0).sum())
```

```python
delivery_performance = {
    "Early": (delivery_analysis["delivery_delay_days"] < 0).sum(),
    "On Time": (delivery_analysis["delivery_delay_days"] == 0).sum(),
    "Late": (delivery_analysis["delivery_delay_days"] > 0).sum()
}

delivery_performance_pct = {
    k: round(v / len(delivery_analysis) * 100, 2)
    for k, v in delivery_performance.items()
}

delivery_performance_pct
```

## 6.6 Delivery Status Classification Adjustment

The initial delivery-status calculation compared the complete delivery timestamp with the estimated delivery timestamp.

This resulted in zero orders being classified as exactly on time because the actual delivery records contain time-of-day information while the estimated delivery field is represented as a date.

To make the business KPI meaningful, delivery performance is therefore evaluated at the calendar-date level:

- **Early:** Actual delivery date is before the estimated date.
- **On Time:** Actual delivery date matches the estimated date.
- **Late:** Actual delivery date is after the estimated date.

This prevents time-of-day differences from incorrectly classifying an order delivered on the estimated calendar date as late.

```python
delivery_analysis["actual_delivery_date"] = (
    delivery_analysis["order_delivered_customer_date"].dt.normalize()
)

delivery_analysis["estimated_delivery_date"] = (
    delivery_analysis["order_estimated_delivery_date"].dt.normalize()
)

delivery_analysis["delivery_delay_days"] = (
    delivery_analysis["actual_delivery_date"]
    - delivery_analysis["estimated_delivery_date"]
).dt.days
```

```python
delivery_analysis["delivery_status"] = delivery_analysis["delivery_delay_days"].apply(
    lambda x: "Early" if x < 0
    else "On Time" if x == 0
    else "Late"
)

delivery_status_counts = delivery_analysis["delivery_status"].value_counts()

delivery_status_pct = (
    delivery_analysis["delivery_status"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print(delivery_status_counts)
print()
print(delivery_status_pct)
```

## 12.1 Creating the Order-Item Analysis Dataset

The order-item level will be used as the foundation for product, category, seller, and revenue analysis.

Orders, customers, products, and sellers will be joined using their respective keys.

Payments and reviews will remain separate at this stage because an order can contain multiple payment or review records. Joining them directly could duplicate order-item records and produce incorrect revenue calculations.

The analysis dataset will preserve the original cleaned datasets while creating a separate DataFrame for business analysis.v

```python
order_item_analysis = (
    order_items_clean
    .merge(
        orders_clean,
        on="order_id",
        how="left"
    )
    .merge(
        customers_clean,
        on="customer_id",
        how="left"
    )
    .merge(
        products_clean[
            [
                "product_id",
                "product_category_name",
                "product_category_english"
            ]
        ],
        on="product_id",
        how="left"
    )
    .merge(
        sellers_clean,
        on="seller_id",
        how="left"
    )
)

print("Analysis dataset shape:", order_item_analysis.shape)

print("\nMissing values in key joined fields:")
print(
    order_item_analysis[
        [
            "order_id",
            "customer_id",
            "product_id",
            "seller_id",
            "product_category_english"
        ]
    ].isnull().sum()
)

print("\nSample:")
order_item_analysis.head()
```

## 12.2 Core Revenue KPIs

Revenue will be calculated from the `price` field in the order-item dataset.

The `price` field represents the item price, while `freight_value` represents the freight/shipping charge. They will be analyzed separately.

The following KPIs will be calculated:

- Total item revenue
- Total freight value
- Total order items
- Unique orders
- Average order value (AOV)

Average order value will be calculated at the order level so that orders containing multiple items are not counted multiple times.

```python
# Total item revenue
total_revenue = order_item_analysis["price"].sum()

# Total freight
total_freight = order_item_analysis["freight_value"].sum()

# Total order items
total_items = len(order_item_analysis)

# Unique orders
unique_orders = order_item_analysis["order_id"].nunique()

# Revenue per order
order_revenue = (
    order_item_analysis
    .groupby("order_id")["price"]
    .sum()
)

average_order_value = order_revenue.mean()

print("CORE REVENUE KPIs")
print("=" * 40)
print(f"Total item revenue: {total_revenue:,.2f}")
print(f"Total freight value: {total_freight:,.2f}")
print(f"Total order items: {total_items:,}")
print(f"Unique orders with items: {unique_orders:,}")
print(f"Average order value: {average_order_value:,.2f}")
```

## 12.3 Core Revenue Findings

The initial revenue analysis shows:

- Total item revenue of **13,591,643.70** across the available order-item records.
- Total freight value of **2,251,909.54**.
- **112,650** order-item records across **98,666** unique orders containing items.

Item revenue and freight value are kept as separate metrics because freight represents shipping charges rather than product sales revenue.

## 12.4 Monthly Revenue Analysis

Monthly revenue will be calculated using the order purchase date and item price.

This analysis will help identify revenue trends, periods of higher or lower sales activity, and potential seasonal patterns.

Revenue will be aggregated by month at the order-item level.

```python
monthly_revenue = (
    order_item_analysis
    .assign(
        order_month=order_item_analysis[
            "order_purchase_timestamp"
        ].dt.to_period("M")
    )
    .groupby("order_month")
    .agg(
        revenue=("price", "sum"),
        orders=("order_id", "nunique"),
        items=("order_id", "count")
    )
    .reset_index()
)

monthly_revenue["order_month"] = (
    monthly_revenue["order_month"].astype(str)
)

monthly_revenue["average_order_value"] = (
    monthly_revenue["revenue"] /
    monthly_revenue["orders"]
)

monthly_revenue.head()
```

```python
print("Number of months:", len(monthly_revenue))

print("\nHighest revenue months:")
print(
    monthly_revenue
    .sort_values("revenue", ascending=False)
    .head(10)
)

print("\nLowest revenue months:")
print(
    monthly_revenue
    .sort_values("revenue")
    .head(10)
)
```

## 14.11 Customer Recency Analysis

Customer recency measures how recently each customer placed an order relative to the latest purchase date available in the dataset.

The most recent purchase date in the dataset will be used as the reference date. This avoids using the current calendar date, which would incorrectly make all historical customers appear inactive.

Recency will be measured in days:

**Recency = Reference Date − Customer's Most Recent Order Date**

A lower recency value indicates a more recent purchase.

```python
reference_date = order_item_analysis[
    "order_purchase_timestamp"
].max().normalize()

customer_recency = (
    order_item_analysis
    .groupby("customer_unique_id")
    .agg(
        last_purchase_date=("order_purchase_timestamp", "max")
    )
    .reset_index()
)

customer_recency["recency_days"] = (
    reference_date -
    customer_recency["last_purchase_date"].dt.normalize()
).dt.days

print("CUSTOMER RECENCY")
print("=" * 45)

print("Reference date:", reference_date)
print("Customers analyzed:", customer_recency["customer_unique_id"].nunique())

print("\nRecency statistics:")
print(
    customer_recency["recency_days"].describe()
)
```

These thresholds are derived from the dataset's observed 25th, 50th, and 75th percentiles rather than being externally defined benchmarks.

```python
customer_recency["recency_segment"] = pd.cut(
    customer_recency["recency_days"],
    bins=[-1, 119, 224, 353, float("inf")],
    labels=[
        "Very Recent",
        "Recent",
        "Less Recent",
        "Distant"
    ]
)

recency_segments = (
    customer_recency
    .groupby("recency_segment", observed=False)
    .agg(
        customers=("customer_unique_id", "count"),
        average_recency_days=("recency_days", "mean")
    )
    .reset_index()
)

recency_segments["customer_share_pct"] = (
    recency_segments["customers"] /
    recency_segments["customers"].sum()
) * 100

recency_segments
```

## 14.18 RFM Scoring Method

RFM scores will be created to compare customers across three dimensions:

- **Recency (R):** More recent purchases receive higher scores.
- **Frequency (F):** Customers with more orders receive higher scores.
- **Monetary (M):** Customers generating higher item revenue receive higher scores.

Different scoring approaches are used because the three variables have different distributions.

### Recency Scoring

Recency will use the observed quartiles:

- 1 = Distant
- 2 = Less Recent
- 3 = Recent
- 4 = Very Recent

A higher Recency score therefore represents more recent customer activity.

### Frequency Scoring

Frequency is highly concentrated because most customers placed only one order.

Therefore, fixed purchase-frequency groups will be used:

- 1 order → Score 1
- 2 orders → Score 2
- 3 orders → Score 3
- 4 orders → Score 4
- 5 or more orders → Score 5

### Monetary Scoring

Monetary value will be scored using revenue quintiles so that customers are divided according to their relative revenue contribution.

- Lowest 20% → Score 1
- Next 20% → Score 2
- Middle 20% → Score 3
- Next 20% → Score 4
- Highest 20% → Score 5

The scores are relative to the observed Olist customer distribution and should not be interpreted as industry benchmarks.

```python
# Recency score
recency_score_map = {
    "Distant": 1,
    "Less Recent": 2,
    "Recent": 3,
    "Very Recent": 4
}

customer_rfm["R_score"] = (
    customer_rfm["recency_segment"]
    .map(recency_score_map)
    .astype(int)
)


# Frequency score
def frequency_score(order_count):
    if order_count == 1:
        return 1
    elif order_count == 2:
        return 2
    elif order_count == 3:
        return 3
    elif order_count == 4:
        return 4
    else:
        return 5


customer_rfm["F_score"] = (
    customer_rfm["orders"]
    .apply(frequency_score)
)


# Monetary score
customer_rfm["M_score"] = pd.qcut(
    customer_rfm["revenue"],
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)


print("RFM SCORE VALIDATION")
print("=" * 50)

print("\nRecency scores:")
print(customer_rfm["R_score"].value_counts().sort_index())

print("\nFrequency scores:")
print(customer_rfm["F_score"].value_counts().sort_index())

print("\nMonetary scores:")
print(customer_rfm["M_score"].value_counts().sort_index())
```

## 14.21 Creating the RFM Score

The individual Recency, Frequency, and Monetary scores will be combined into a three-digit RFM score.

The score follows the format:

**RFM = Recency + Frequency + Monetary**

For example:

- `415` = very recent purchase, one order, high monetary value
- `154` = distant purchase, one order, high monetary value
- `555` = very recent, highly frequent, high monetary value

The three individual scores will be retained alongside the combined RFM score so that customer behavior can be interpreted across all dimensions.m

```python
customer_rfm["RFM_score"] = (
    customer_rfm["R_score"].astype(str) +
    customer_rfm["F_score"].astype(str) +
    customer_rfm["M_score"].astype(str)
)

print("RFM SCORE SAMPLE")
print("=" * 50)

print(
    customer_rfm[
        [
            "customer_unique_id",
            "recency_days",
            "orders",
            "revenue",
            "R_score",
            "F_score",
            "M_score",
            "RFM_score"
        ]
    ].head(10)
)

print("\nUnique RFM scores:", customer_rfm["RFM_score"].nunique())
```

## 14.22 RFM Customer Segmentation

The individual RFM scores will be converted into descriptive customer segments.

The segmentation uses the customer's Recency, Frequency, and Monetary scores together.

The segments are designed to describe observed customer behavior rather than assign a subjective ranking to customers.

The following rules will be used:

- **High Value:** High recency, frequency, and monetary scores.
- **Recent Customers:** High recency score but lower frequency.
- **Repeat Customers:** Higher frequency score indicating multiple purchases.
- **High Spending:** High monetary score with lower or moderate frequency.
- **Low Engagement:** Low recency score combined with low frequency.
- **Other:** Customers whose RFM combination does not fall into the above descriptive groups.

The original RFM scores will be retained so the segmentation rules remain transparent.

```python
def assign_rfm_segment(row):
    r = row["R_score"]
    f = row["F_score"]
    m = row["M_score"]

    if r >= 3 and f >= 3 and m >= 3:
        return "High Value"

    elif r >= 3 and f <= 2:
        return "Recent Customers"

    elif f >= 3:
        return "Repeat Customers"

    elif m >= 4:
        return "High Spending"

    elif r <= 2 and f <= 2:
        return "Low Engagement"

    else:
        return "Other"


customer_rfm["rfm_segment"] = (
    customer_rfm.apply(assign_rfm_segment, axis=1)
)

rfm_segments = (
    customer_rfm
    .groupby("rfm_segment")
    .agg(
        customers=("customer_unique_id", "count"),
        average_recency=("recency_days", "mean"),
        average_orders=("orders", "mean"),
        total_revenue=("revenue", "sum"),
        average_revenue=("revenue", "mean")
    )
    .reset_index()
)

rfm_segments["customer_share_pct"] = (
    rfm_segments["customers"] /
    rfm_segments["customers"].sum()
) * 100

rfm_segments
```
